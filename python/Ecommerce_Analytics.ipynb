```python
orders_customers = orders.merge(
    customers,
    on="Customer_ID",
    how="left",
    validate="many_to_one"
)

print("Orders + Customers shape:", orders_customers.shape)
display(orders_customers.head())
```

```python
analysis_df = orders_customers.merge(
    products,
    on="Product_ID",
    how="left",
    validate="many_to_one"
)

print("Orders + Customers + Products shape:", analysis_df.shape)
display(analysis_df.head())
```

```python
return_summary = (
    returns.groupby("Order_ID")
    .agg(
        Return_Count=("Return_ID", "count"),
        Return_Reason=("Return_Reason", "first")
    )
    .reset_index()
)

display(return_summary.head())
```

```python
analysis_df = analysis_df.merge(
    return_summary,
    on="Order_ID",
    how="left",
    validate="many_to_one"
)

analysis_df["Return_Count"] = analysis_df["Return_Count"].fillna(0)

analysis_df["Returned"] = np.where(
    analysis_df["Return_Count"] > 0,
    "Yes",
    "No"
)

display(analysis_df.head())
```

```python
analysis_df["Revenue"] = (
    analysis_df["Quantity"]
    * analysis_df["Selling_Price"]
    * (1 - analysis_df["Discount"])
)
```

```python
analysis_df["Cost"] = (
    analysis_df["Quantity"]
    * analysis_df["Cost_Price"]
)
```

```python
analysis_df["Profit"] = (
    analysis_df["Revenue"]
    - analysis_df["Cost"]
    - analysis_df["Shipping_Cost"]
)
```

```python
analysis_df["Profit_Margin_Percent"] = np.where(
    analysis_df["Revenue"] != 0,
    (analysis_df["Profit"] / analysis_df["Revenue"]) * 100,
    0
)
```

```python
total_orders = analysis_df["Order_ID"].nunique()
total_quantity = analysis_df["Quantity"].sum()
total_revenue = analysis_df["Revenue"].sum()
total_profit = analysis_df["Profit"].sum()

average_order_value = total_revenue / total_orders
profit_margin = (total_profit / total_revenue) * 100

print("===== OVERALL KPIs =====")
print("Total Orders:", total_orders)
print("Total Quantity:", total_quantity)
print("Total Revenue:", round(total_revenue, 2))
print("Total Profit:", round(total_profit, 2))
print("Average Order Value:", round(average_order_value, 2))
print("Profit Margin:", round(profit_margin, 2), "%")
```

```python
customer_last_order = (
    orders[orders["Order_Status"] != "Cancelled"]
    .groupby("Customer_ID")["Order_Date"]
    .max()
    .reset_index()
)

customer_last_order["Days_Since_Last_Order"] = (
    orders["Order_Date"].max()
    - customer_last_order["Order_Date"]
).dt.days

average_recency = customer_last_order[
    "Days_Since_Last_Order"
].mean()

at_risk_customers = customer_last_order[
    customer_last_order["Days_Since_Last_Order"] > average_recency
].merge(
    customers[["Customer_ID", "Customer_Name", "Region"]],
    on="Customer_ID",
    how="left"
).sort_values(
    "Days_Since_Last_Order",
    ascending=False
)

display(at_risk_customers.head(20))
```

```python
monthly_sales["Total_Revenue"] = monthly_sales["Total_Revenue"].astype(float)

monthly_sales["Previous_Month_Revenue"] = (
    monthly_sales["Total_Revenue"].shift(1)
)

monthly_sales["MoM_Growth_Percent"] = (
    (
        monthly_sales["Total_Revenue"]
        - monthly_sales["Previous_Month_Revenue"]
    )
    / monthly_sales["Previous_Month_Revenue"]
) * 100

display(monthly_sales)
```

```python
analysis_date = analysis_df["Order_Date"].max()

rfm = analysis_df[
    analysis_df["Order_Status"] != "Cancelled"
].groupby("Customer_ID").agg(
    Recency=("Order_Date", lambda x: (analysis_date - x.max()).days),
    Frequency=("Order_ID", "nunique"),
    Monetary=("Revenue", "sum")
).reset_index()

display(rfm.head())
```

```python
rfm["R_Score"] = pd.qcut(
    rfm["Recency"],
    5,
    labels=[5, 4, 3, 2, 1]
)

rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
)

rfm["M_Score"] = pd.qcut(
    rfm["Monetary"],
    5,
    labels=[1, 2, 3, 4, 5]
)
```

```python
rfm["RFM_Score"] = (
    rfm["R_Score"].astype(int)
    + rfm["F_Score"].astype(int)
    + rfm["M_Score"].astype(int)
)

display(rfm.head(10))
```

```python
def segment_customer(row):
    score = row["RFM_Score"]

    if score >= 13:
        return "Champions"
    elif score >= 11:
        return "Loyal Customers"
    elif score >= 9:
        return "Potential Loyalists"
    elif score >= 7:
        return "New Customers"
    elif score >= 5:
        return "At Risk"
    elif score >= 4:
        return "Can't Lose Them"
    else:
        return "Lost Customers"

rfm["Customer_Segment"] = rfm.apply(
    segment_customer,
    axis=1
)

display(rfm.head(10))
```

```python
segment_summary = (
    rfm.groupby("Customer_Segment")
    .agg(
        Customer_Count=("Customer_ID", "count"),
        Total_Revenue=("Monetary", "sum"),
        Average_Revenue=("Monetary", "mean")
    )
    .reset_index()
    .sort_values(
        "Total_Revenue",
        ascending=False
    )
)

display(segment_summary)
```

```python
top_rfm_segment = segment_summary.iloc[0]

print(
    "Highest revenue RFM segment:",
    top_rfm_segment["Customer_Segment"]
)

print(
    "Revenue:",
    round(top_rfm_segment["Total_Revenue"], 2)
)
```
